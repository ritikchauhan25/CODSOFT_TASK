**Visualization Style**

```python
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)
```

**Customer Churn Bar Chart**

```python
churn_counts = df["Churn"].value_counts().reindex(["No", "Yes"])

churn_counts.plot(kind="bar", color=["steelblue", "coral"])

plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Customer Churn Pie Chart**

```python
churn_counts = df["Churn"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Percentage")
plt.show()
```

**Customer Tenure Histogram**

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="tenure", bins=30, kde=True)

plt.title("Distribution of Customer Tenure")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

**Contract Type vs Churn**

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

**Monthly Charges Box Plot**

```python
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")

plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Monthly Charges")
plt.tight_layout()
plt.show()
```

**Tenure vs Monthly Charges Scatter Plot**

```python
sns.scatterplot(
    data=df,
    x="tenure",
    y="MonthlyCharges",
    hue="Churn",
    alpha=0.6
)

plt.title("Tenure vs Monthly Charges")
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges")
plt.tight_layout()
plt.show()
```

**Internet Service vs Churn**

```python
sns.countplot(data=df, x="InternetService", hue="Churn")

plt.title("Customer Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

**Payment Method vs Churn**

```python
sns.countplot(data=df, x="PaymentMethod", hue="Churn")

plt.title("Customer Churn by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

**Average Monthly Charges by Churn**

```python
avg_charges = df.groupby("Churn")["MonthlyCharges"].mean()

avg_charges.plot(kind="bar", color=["steelblue", "coral"])

plt.title("Average Monthly Charges by Churn")
plt.xlabel("Churn Status")
plt.ylabel("Average Monthly Charges")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(avg_charges)
```
